```python
# numpy is useful for numeric operations if needed during analysis.
import numpy as np
# pandas is used to load and work with the dataset in table form.
import pandas as pd
# seaborn is a visualization library. It is imported here because it is commonly used with matplotlib.
import seaborn as sns
# matplotlib.pyplot is used to create histograms, boxplots and figure layouts.
import matplotlib.pyplot as plt
# This line shows graphs directly inside the notebook output.
%matplotlib inline
```

```python
# read_csv() loads the Iris dataset into a DataFrame.
df = pd.read_csv("iris.csv")

# We keep the original Iris feature names so the notebook is easier to understand.
# The first column is Id, the next four columns are numeric flower measurements,
# and the last column is the categorical species label.
df.columns = ['Id', 'SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species']

# head() shows the first 5 rows so we can understand the data structure.
df.head()
```

```python
# len(list(df)) counts how many columns are present in the dataset.
# This helps us quickly understand the size of the dataset structure.
column = len(list(df))
column
```

```python
# info() shows column names, number of non-null values and data types.
# We use it to identify which columns are numeric and which are categorical.
df.info()
```

```python
# unique() gives the different class names present in the target column.
# We use Species here because it stores the flower category labels.
df['Species'].unique()
```

```python
# This lists the features and their types clearly as required in the practical.
print('Feature Types in Iris Dataset:')
print('1. Id - Integer / Numeric identifier')
print('2. SepalLengthCm - Numeric')
print('3. SepalWidthCm - Numeric')
print('4. PetalLengthCm - Numeric')
print('5. PetalWidthCm - Numeric')
print('6. Species - Nominal / Categorical')
```

```python
# subplots(2, 2) creates four graph areas in one figure.
# We use this because there are four numeric features to compare together.
# figsize sets the width and height of the full figure.
fig, axes = plt.subplots(2, 2, figsize=(16, 8))

# hist() creates a histogram.
# In every histogram, the x-axis shows feature values and the y-axis shows frequency.
# Frequency means how many flowers have values in a particular range.
# This helps us understand spread, concentration and shape of the distribution.
axes[0,0].set_title('Distribution of SepalLengthCm')
axes[0,0].hist(df['SepalLengthCm'])

axes[0,1].set_title('Distribution of SepalWidthCm')
axes[0,1].hist(df['SepalWidthCm'])

axes[1,0].set_title('Distribution of PetalLengthCm')
axes[1,0].hist(df['PetalLengthCm'])

axes[1,1].set_title('Distribution of PetalWidthCm')
axes[1,1].hist(df['PetalWidthCm'])

# tight_layout() adjusts spacing so titles and graphs do not overlap.
plt.tight_layout()
# show() displays the full figure output.
plt.show()
```

```python
# boxplot() is used because it clearly shows minimum, maximum, median and possible outliers.
# Median is the middle value of the data.
# Outliers are values that are unusually far from the rest of the data.
# We use all numeric feature columns together so comparison becomes easy.
# The x-axis shows feature names and the y-axis shows their numeric values.
# This graph helps us compare spread and check which feature may contain outliers.
df.boxplot(column=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'], figsize=(10, 5))
# The title explains that this graph is the boxplot of all feature columns.
plt.title('Boxplot of All Features')
# We rotate x-axis labels so the feature names are easier to read.
plt.xticks(rotation=45)
# show() displays the graph output.
plt.show()
```

```python
# describe() gives summary statistics such as count, mean, standard deviation, minimum and maximum.
# We use it to support the graph observations with numeric values.
df.describe()
```

```python
# feature_cols stores the names of the numeric columns we want to plot one by one.
feature_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

# This loop creates a separate histogram for each numeric feature.
for col in feature_cols:
    # figure(figsize=(8, 4)) creates one graph area of width 8 and height 4.
    plt.figure(figsize=(8, 4))
    # hist() is used to show distribution of one feature.
    # We take the current feature column on the x-axis because we want to study its values.
    # bins=10 means the values are divided into 10 intervals or groups.
    # color sets the fill color of bars and edgecolor sets the bar border color.
    # The y-axis shows frequency, meaning how many data values fall in each interval.
    # This graph shows whether the feature values are concentrated, spread out, symmetric or skewed.
    plt.hist(df[col], bins=10, color='lightgreen', edgecolor='black')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
# This loop creates a separate boxplot for each numeric feature.
for col in feature_cols:
    # figure(figsize=(8, 4)) creates a separate graph area for each feature.
    plt.figure(figsize=(8, 4))
    # boxplot(column=[col]) draws the box plot of one feature only.
    # The x-axis shows the selected feature name and the y-axis shows its values.
    # The box shows the middle spread of the data and the line inside the box shows the median.
    # The whiskers show the normal range of the data, while points far away may be outliers.
    # This graph helps us check spread, central tendency and unusual values in each feature.
    df.boxplot(column=[col])
    plt.title(f'Boxplot of {col}')
    plt.ylabel(col)
    plt.grid(True)
    plt.show()
```

```python
# Observations and inferences from the histograms and boxplots.
print('Observations and Inferences:')
print('1. SepalLengthCm and SepalWidthCm show moderate spread in their distributions.')
print('2. PetalLengthCm and PetalWidthCm show clearer grouping, which suggests stronger separation among Iris species.')
print('3. Petal-based features vary more distinctly than sepal-based features, so their distributions are easier to compare visually.')
print('4. From the boxplots, SepalWidthCm appears to contain some outliers because a few values lie away from the main range.')
print('5. The boxplots help compare spread, median and unusual values across all numeric features.')
```
