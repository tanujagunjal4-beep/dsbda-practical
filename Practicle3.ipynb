```python
'''Descriptive Statistics 
• Compute mean, median, min, max, variance and standard deviation  
• Group data by categorical variable  
• Analyze Iris dataset statistics (species-wise comparison) '''
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Iris.csv")
```

```python
df.head()
```

```python
# Display dataset information
df.info()
```

```python
#Compute Mean
df.mean(numeric_only=True)
```

```python
#Compute Median
df.median(numeric_only=True)
```

```python
#Compute Minimum Values
df.min(numeric_only=True)
```

```python
#Compute Maximum Values
df.max(numeric_only=True)
```

```python
# Compute Variance
df.var(numeric_only=True)
```

```python
#Compute Standard Deviation
df.std(numeric_only=True)
```

```python
#Group Data by Species
group = df.groupby('Species')

group.size()
```

```python
#Species-wise Mean Comparison
group.mean(numeric_only=True)
```

```python
#Species-wise Median Comparison
group.median(numeric_only=True)
```

```python
#Species-wise Minimum Values
group.min(numeric_only=True)
```

```python
#Species-wise Maximum Values
group.max(numeric_only=True)
```

```python
#Species-wise Standard Deviation
group.std(numeric_only=True)
```
