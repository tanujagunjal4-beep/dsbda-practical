```python
df["math_score"]=df["math_score"].fillna(df["math_score"].mean())
```

```python
df["math_score"].isnull().sum()
```

```python
df["Placement_score"]=df["Placement_score"].dropna()
```

```python
df["Placement_score"]=df["Placement_score"].fillna(df["Placement_score"].mean())
```

```python
df["Reading_score"]=df["Reading_score"].fillna(df["Reading_score"].mean())
```

```python
df["Writing_score"]=df["Writing_score"].fillna(df["Writing_score"].mean())
```

```python
df["Club_Join_Date"]=df["Club_Join_Date"].fillna(df["Club_Join_Date"].mode())
```

```python
df.isnull().sum()
```

```python
df.boxplot(column=["math_score", "Reading_score", "Writing_score", "Placement_score"])
plt.title("Student Performance")
plt.xlabel("Score")
plt.ylabel("Range")
```
